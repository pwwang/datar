1. dr.slice(): allows selecting rows by their positions (indices).
   + Basic usage: dr.slice_(index1, index2, ...)
   + Range indexing: dr.slice_(dr.c[start:end]) to select rows from start to end-1.
  
<br>

2. dr.slice_min(): allows selecting rows with the smallest values in a specified column.

<br>

3. dr.slice_max(): allows selecting rows with the largest values in a specified column.

```python
import datar.all as dr
from datar import f
import pandas as pd
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")
pd.set_option("display.width", 200)

# Get the path object pointing to the ``notebooks`` directory that contains *.csv files 
data_dir = next(Path("/home").rglob("*/notebooks/*.csv")).parent
```

```python
df_baseball = pd.read_csv(
    filepath_or_buffer=data_dir/"baseball.csv",
    dtype={
        "Team": "category",
        "Position": "category",
        "PosCategory": "category"
    }
)

print(df_baseball.head())
```

# <span style="color:#1E90FF">1. dr.slice(...)</span>

```python
from pipda import register_verb

dr.slice = register_verb(func=dr.slice_)

'''
dr.slice_() to avoid conflict with Python built-in function slice()
=> Use registter_verb to bring it back as dr.slice()
'''
```

```python
#################
## Basic usage ##
#################

print(
    df_baseball
    >> dr.slice(0)
)
```

```python
print(
    df_baseball
    >> dr.slice(0, 2, 5, 9)
)
```

```python
print(
    df_baseball
    >> dr.slice(-1, -2)
)
```

```python
####################
## Range indexing ##
####################

print(
    df_baseball
    >> dr.slice(dr.c[0:4])
)
```

```python
print(
    df_baseball
    >> dr.slice(dr.c[:3])
)
```

```python
print(
    df_baseball
    >> dr.slice(dr.c[1010:])
)
```

```python
print(
    df_baseball
    >> dr.slice(dr.c[0:10:2])
)
```

# <span style="color:#1E90FF">2. dr.slice_min(...)</span>
allows selecting rows with the smallest values in a specified column

```python
print(
    df_baseball
    >> dr.slice_min(f.Age, n=3) # Select 3 rows with the smallest Age
)
```

```python
print(
    df_baseball
    >> dr.slice_min(f.Age, n=5, with_ties=True) # Include ties
)
```

# <span style="color:#1E90FF">3. dr.slice_max(...)</span>
allows selecting rows with the largest values in a specified column.

```python
print(
    df_baseball
    >> dr.slice_max(f.Height, n=3) # Select 3 rows with the largest Height
)
```

```python
print(
    df_baseball
    >> dr.slice_max(f.Height, n=5, with_ties=True) # Include ties
)
```
