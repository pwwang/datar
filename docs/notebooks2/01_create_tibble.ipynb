## ```tibble()``` function from DataR library creates a DataFrame in a more user-friendly way compared to the traditional pandas DataFrame constructor.

--------------------------------------------------------------------------------

1. Create dataframe using ```datar.tibble.tibble()```

2. Convert Pandas DataFrame to DataR Tibble

````python
import datar.all as dr
import pandas as pd
````

# <span style="color:#1E90FF">1. Create dataframe using datar.tibble.tibble()</span>

````python
tb = dr.tibble(
    x = [1, 2, 3],
    y = ["a", "b", "c"],
    z = [True, False, True]
)

print(type(tb))
````

````python
print(tb)
````

# <span style="color:#1E90FF">2. Convert Pandas DataFrame to DataR Tibble</span>

````python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': ['x', 'y', 'z'],
    'C': [True, False, True]
})

# convert from Pan
tb_from_df = dr.tibble(df)

print(tb_from_df)
````
